## Print examples

```python
from PIL import Image
import matplotlib.pyplot as plt

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 10))

ax1.imshow(Image.open("data_original/chest_xray/chest_xray/train/NORMAL/IM-0115-0001.jpeg"), cmap="gray")
ax1.set_title("Normal")

ax2.imshow(Image.open("data_original/chest_xray/chest_xray/train/NORMAL/IM-0117-0001.jpeg"), cmap="gray")
ax2.set_title("Normal")

ax3.imshow(Image.open("data_original/chest_xray/chest_xray/train/PNEUMONIA/person1_bacteria_1.jpeg"), cmap="gray")
ax3.set_title("Pneumonia")

ax4.imshow(Image.open("data_original/chest_xray/chest_xray/train/PNEUMONIA/person1_bacteria_2.jpeg"), cmap="gray")
ax4.set_title("Pneumonia")

plt.tight_layout()
plt.show()
```
